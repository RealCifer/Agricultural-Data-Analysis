## GENERATE 2018–2025 (ALL PREDICTED)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df = pd.read_csv('../data/tamilnadu_rice_2000_2017.csv',
                 usecols=['Year','Yield_kg_per_ha'])

df = df.dropna()

X = df[['Year']]
y = df['Yield_kg_per_ha']

model = LinearRegression().fit(X, y)

future_years = np.arange(2018, 2026).reshape(-1,1)
pred = model.predict(future_years)

future_df = pd.DataFrame({
    'Year': future_years.flatten(),
    'Yield_kg_per_ha': pred
})

print("Predicted Data (2018–2025):")
print(future_df)
```
